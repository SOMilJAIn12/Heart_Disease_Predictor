```python
df_copy_c=df_copy
```

```python
cat_col=["cp","restecg","thal"]
df_copy_c=pd.get_dummies(df_copy_c,columns=cat_col,drop_first=True)
```

```python
X=df_copy_c.drop("num",axis=1)
y=df_copy_c["num"]
```

```python
X,y
```

```python
from collections import Counter
Counter(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.30,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
regression=LogisticRegression()
from sklearn.model_selection import RandomizedSearchCV
penalty=["l1","l2","elasticnet"]
solver=["newton-cg","lbfgs","sag","saga","liblinear"]
c_score=[100,10,1,.1,.01]
params=dict(penalty=penalty,solver=solver,C=c_score)
params
```

```python
random=RandomizedSearchCV(estimator=regression,param_distributions=params,cv=5,scoring="accuracy")
random.fit(X_train_scaled,y_train)
```

```python
y_pred=random.predict(X_test_scaled)
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
ac=accuracy_score(y_test,y_pred)
cm=confusion_matrix(y_test,y_pred)
cr=classification_report(y_test,y_pred)
ac
```
